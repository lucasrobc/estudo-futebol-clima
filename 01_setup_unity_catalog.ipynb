# 01 - Setup do Unity Catalog
Cria o schema e os volumes do projeto.

| Objeto | Nome completo | Para que serve |
|---|---|---|
| Schema | `workspace.fc` | Agrupa tudo do projeto |
| Volume | `workspace.fc.raw` | JSON bruto das APIs (landing zone) |
| Volume | `workspace.fc.output` | Dataset final em Parquet |

```python
%sql
CREATE SCHEMA IF NOT EXISTS workspace.fc
COMMENT 'Desafio Python - Projeto 1: pipeline futebol x clima';
```

```python
%sql
CREATE VOLUME IF NOT EXISTS workspace.fc.raw
COMMENT 'Respostas JSON brutas das APIs';
```

```python
%sql
CREATE VOLUME IF NOT EXISTS workspace.fc.output
COMMENT 'Camada final em Parquet (Snappy), particionada por ano/mes';
```
