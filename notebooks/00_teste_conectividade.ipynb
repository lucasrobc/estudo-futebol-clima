# Teste de conectividade
Teste se as APIs respondem na versão Free Edition

```python
import requests

urls = {
    "football-data.org": "https://api.football-data.org/v4/competitions",
    "open-meteo": (
        "https://archive-api.open-meteo.com/v1/archive"
        "?latitude=-23.55&longitude=-46.63"
        "&start_date=2026-01-01&end_date=2026-01-03"
        "&daily=temperature_2m_max"
    ),
}

for nome, url in urls.items():
    try:
        resposta = requests.get(url, timeout=20)
        print(f"OK {nome}: HTTP {resposta.status_code}")
    except requests.exceptions.RequestException as erro:
        print(f"ERRO {nome}: bloqueado ou fora do ar -> {type(erro).__name__}")
```

```python
resposta = requests.get("https://api.football-data.org/v4/competitions/BSA/matches", timeout=20)
print(resposta.status_code, "->", resposta.json().get("message"))
```

```python
resposta = requests.get(urls["open-meteo"], timeout=20)
print("Status:", resposta.status_code)
print("Corpo:", resposta.text[:300])
print("Retry-After:", resposta.headers.get("Retry-After"))
```
